```python
def number():
    n=int(input("enter a number"))
    count=0
    for i in str(n):
        count+=1
        if count==9:
            print(i)
number()            
```

```python
5. Write a python program to Generate N Digit Numbers with Alternating Even and Odd Digits
Generate numbers with patterns like: 2, 3, 2, 3... or 1, 4, 1... where even and odd digits alternate.
```
