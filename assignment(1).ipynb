```python
def sum():
 n= int(input("enter a number"))
 total =0
 sign=1
 for i in str(n):
     total+=sign*int(i)
     sign =-sign
     print(total)
sum()    
```

```python
10.Write a python function to Count How Many Times Digit Changes (Even to Odd or Odd to Even)

e.g. 123456 → 1(odd)→2(even)→3(odd)→4(even)... → 5 changes
```

```python
n= int(input("enter a number"))
count=0
for i in str(n):
    d1 = int(n[i])
    d2 = int(n[i + 1])
    if (d1 % 2 == 0 and d2 % 2 != 0) or (d1 % 2 != 0 and d2 % 2 == 0):
            count += 1

return count
    
             
```
